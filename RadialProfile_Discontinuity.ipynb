Blast Wave Radial Profile 2003

```python
import matplotlib.pyplot as plt
import numpy as np
from astropy.io import fits
```

```python
BWRProfile_file = 'BW2003RadAngProfile.fits'
```

```python
with fits.open(BWRProfile_file, ignore_missing_end=True) as hdu_list:
    hdu_list.verify('silentfix')
    hist_dataBlastWave = hdu_list[1].data

# Grab the radius values
radiusBW = hist_dataBlastWave['RMID']

# Need to convert from pixels to arcsec using plate scale of 0.492"
# radius2003 = radius2003 * 0.492


# Grab the angle values
angleBW = hist_dataBlastWave['ROTANG']
angleBW = angleBW[:,0]


# Grab the flux values given in photons/cm**2/pixel**2/s
fluxBW = hist_dataBlastWave['SUR_FLUX']

# Convert flux to arcsec
# flux2003 = flux2003 * 4.1311388723641

# Grab the flux error values given in pixels**-2
# fluxerrBW = hist_dataBlastWave['SUR_FLUX_ERR']
# Convert to arcsec**-2
# fluxerr2003 = fluxerr2003 * 4.1311388723641
```

```python
radlist = []
fluxlist = []

for theta, rad, flux in zip(angleBW, radiusBW, fluxBW):
    if theta == 0.0:
        radlist.append(rad)
        fluxlist.append(flux)
```

```python
plt.plot(radlist, fluxlist)
```
